```python
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import matplotlib.pyplot as plt

# 1. Model create
model_dt = DecisionTreeRegressor(max_depth=3)

# 2. Train model (IMPORTANT!)
model_dt.fit(x_train, y_train)

# 3. Prediction
pred_dt = model_dt.predict(x_test)

# 4. Metrics
dt_mse = mean_squared_error(y_test, pred_dt)
dt_mae = mean_absolute_error(y_test, pred_dt)
dt_r2 = r2_score(y_test, pred_dt)

print("MSE:", dt_mse)
print("MAE:", dt_mae)
print("R2 Score:", dt_r2)

# 5. Tree Graph (Root node + Leaf node)
plt.figure(figsize=(15,10))
plot_tree(model_dt,
          feature_names=x_train.columns,   # if pandas dataframe
          filled=True)

plt.title("Decision Tree Regression")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# =========================
# MODEL COMPARISON TABLE
# =========================

results = pd.DataFrame({

    'Model': [
        'Linear Regression',
        'Decision Tree',
        'Random Forest'
    ],

    'MAE': [
        lg_MAE,
        dt_mae,
        rf_mae
    ],

    'MSE': [
        lg_MSE,
        dt_mse,
        rf_mse
    ],

    'R2 Score': [
        lg_R2,
        R2,
        rf_r2
    ],

    'Accuracy': [
        lg_acc,
        dt_acc,
        rf_acc
    ]

})

print(results)

# =========================
# BAR GRAPH
# =========================

metrics = ['MAE', 'MSE', 'R2 Score', 'Accuracy']

x = results['Model']

plt.figure(figsize=(12,6))

for metric in metrics:
    plt.bar(x, results[metric], alpha=0.7, label=metric)

plt.ylabel('Score')

plt.title('Machine Learning Model Comparison')

plt.legend()

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()

```
